So the first table I will be taking a look at is the Olist_geolocation table to try and find if there is any correlation between customers and a specific area.

To do this we'll use the query below.

```python
WITH state_counts AS (
  SELECT
    UPPER(TRIM(geolocation_state)) AS cleaned_state,
    COUNT(*) AS total_entries
  FROM
    willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Geolocation
  GROUP BY
    1
)
SELECT
  cleaned_state,
  CASE cleaned_state
    WHEN 'SP' THEN 'São Paulo'
    WHEN 'RJ' THEN 'Rio de Janeiro'
    WHEN 'MG' THEN 'Minas Gerais'
    WHEN 'RS' THEN 'Rio Grande do Sul'
    WHEN 'PR' THEN 'Paraná'
    ELSE 'Other Brazilian State'
  END AS state_full_name,
  ROUND( total_entries * 100 / SUM(total_entries) OVER(),  2) AS percentage_of_total
FROM
  state_counts
ORDER BY total_entries DESC
```

So to break down this code first I'll explain what the idea is. I want to compare the amount of orders to state in which the orders were made to see if a particular state contains a higher percentage of customer which would then let the company know where to best launch their marketing efforts.

## Portion 2

  The next trend I want to explore is the customer_id's number of purchases to see if the company has a high end customer that makes a large number of purchases or possibly even multiple high end customer.

  To do this we'll be using the SQL query below.

```python
SELECT
  DISTINCT customers.customer_unique_id AS unique_customer_id,
  COUNT(order_data.order_id) AS items_bought
FROM
  willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Order_Dataset AS order_data
INNER JOIN
  willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Customers AS customers ON order_data.customer_id = customers.customer_id
GROUP BY
  unique_customer_id
ORDER BY
  items_bought DESC
```

  The first thing I need to do is choose the table I'll be pulling from in the FROM clause which in this case is willy-worldstar.Brazilian_E_Commerce_Data_set.Olist_Order_Dataset.

  Which you can see with this line of code below.

  willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Order_Dataset AS order_data

  Next I will give the alias order_data to the FROM clause using the AS keyword.

Next because we are wanting to compare the number of purchases with the customer_id we'll need to bring another table in to the query and to do this I'll be using the INNER JOIN clause. You can see this in the code line below.

```python
INNER JOIN
  willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Customers AS customers ON order_data.customer_id = customers.
```

To do this the INNER JOIN is going to use the customer column on the Olist set and match it to the customer_id column on the order_data set. While we're at it we'll give the column willy-worldstar-Brazilian_E_Commerce_Data_Set.Olist_Customers the alias customers to make it easier to reference.

After that we will move to the SELECT clause and choose which columns we want to display in our table. To do this I'll use the code below.

```python
SELECT
  DISTINCT customers.customer_unique_id AS unique_customer_id,
  COUNT(order_data.order_id) AS items_bought
```

So to break this down first we are going to select the column customer_unique_id from the customers table and give it the alias unique_customer_id to make it easier to reference. After we do this we are going to apply the DISTINCT operator  to ensure that the data is clean and there is no duplicates.

  One very important thing to note is that in the dataset there is 2 columns in regards to customer_id. These 2 columns are customer_id and unique_customer_id. The standard customer_id which a new one is generated every time a purchase is made. So if we want to ensure we are finding the number of times a customer has repeatedly made purchases we need to make sure that we use the customer_unique_id column and then apply the DISTINCT operator to make sure there is not duplicates of the unique_customer_id records.

After this we want to count the number of times each customer places and order and to do this we use the code below.


```python
COUNT(order_data.order_id) AS items_bought
```

  After we use COUNT to count up all of the orders placed via their id's and give it the Alias items_bought.

  Next we have to include a GROUP BY clause due to the aggregation used in the SELECT clause. In the group by clause we will group by unique_customer_id so that our table will by ordered by customer id's. You can see this in the code below.

```python
GROUP BY
  unique_customer_id
```

  Finally we will need to add an ORDER BY clause to ensure that we see the largest number of orders first. To do this we will order by items_bought and we will add DESC before items_bought so that the largest number of items_bought per customer_id will be displayed first.

  The final thing we will need to do to complete this query is to add an ORDER BY clause. This is because we want our results table to be displayed by a particular column, which in this case will be items_bought. Next we will add the function DESC before items_bought so that the results would be organized by the highest number of items bought first. You can see this in the code block below.

```python
ORDER BY
  items_bought DESC
```

After that we will take a look at the table our query produces and what trends it shows us.

```python
%%bigquery --project willy-worldstar
SELECT
  DISTINCT customers.customer_unique_id AS unique_customer_id,
  COUNT(order_data.order_id) AS items_bought
FROM
  willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Order_Dataset AS order_data
INNER JOIN
  willy-worldstar.Brazilian_E_Commerce_Data_Set.Olist_Customers AS customers ON order_data.customer_id = customers.customer_id
GROUP BY
  unique_customer_id
ORDER BY
  items_bought DESC
```

 What can be seen from our table is that there isn't really any particular customers that purchase a large number of items, however there a few repeat customers. From this data we know there are not very repeat customers and that their customer loyalty  is weak and could use improvement.
